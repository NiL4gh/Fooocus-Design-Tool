# 🎨 Fooocus Designer 2.0
**AI-Powered Graphic Design Asset Generator**

Built on RunDiffusion/Juggernaut-XL-v9 (SDXL) with baked LoRAs for high-quality graphic design asset generation.

### Features:
- ⚡ **Dual-Speed Mode**: ⚡ Fast (~3s SDXL-Lightning) vs 🎯 Master (~15s Juggernaut XL)
- 🎯 **Design Categories & Baked LoRAs** (Silhouette, Flat Vector, Stickers, Logos, Patterns, Posters)
- 🎨 **Fooocus Style Engine**: Curated commercial styles with one-click multi-selection
- 📥 **Drop-to-Load**: Drop generated PNG to restore exact settings
- 🎨 **Curated Color Palettes**: 8 commercial palettes auto-populating swatches
- 🔲 **Transparent PNG Background Removal** (via rembg)
- ✏️ **Vector Mode** (lossless SVG conversion via StarVector-1B)
- 🔄 **Multi-Seed Variations & Mockup Studio**
